```python
from IPython.display import Audio, display
from gtts import gTTS
from ultralytics import YOLO

# 1. Carrega o modelo YOLOv8 versão Nano (rápido e leve)
model = YOLO('yolov8n.pt')


# 2. Função de voz para o sensor inteligente
def emitir_alerta_voz(texto):
  tts = gTTS(text=texto, lang='pt', slow=False)
  tts.save('alerta.mp3')
  display(Audio('alerta.mp3', autoplay=True))
```

```python
# Mapeamento das classes do dia a dia
TRADUCAO = {
    'person': 'uma pessoa',
    'chair': 'uma cadeira',
    'couch': 'um sofá',
    'dining table': 'uma mesa',
    'bottle': 'uma garrafa',
    'cup': 'um copo',
    'cell phone': 'um celular',
    'laptop': 'um notebook',
    'backpack': 'uma mochila',
    'car': 'um veículo',
    'dog': 'um cão',
    'cat': 'um gato',
}
```

```python
# Baixa uma imagem de teste para o sensor analisar
!wget -O teste.jpg https://images.unsplash.com/photo-1517841905240-472988babdf9?w=600

def processar_sensor_visao(caminho_imagem):
    # Executa a detecção de objetos com o YOLOv8
    resultados = model(caminho_imagem)

    # Conjunto para armazenar os itens detectados de forma única
    itens_detectados = set()

    for r in resultados:
        for c in r.boxes.cls:
            nome_classe = model.names[int(c)]
            # Se a classe estiver no nosso dicionário de tradução, adiciona
            if nome_classe in TRADUCAO:
                itens_detectados.add(TRADUCAO[nome_classe])

    # Constrói a mensagem de alerta baseada nos itens encontrados
    if itens_detectados:
        lista_itens = ", ".join(itens_detectados)
        mensagem = f"Atenção! No ambiente eu detectei: {lista_itens}."
    else:
        mensagem = "Nenhum objeto conhecido foi detectado."

    print(f"Mensagem gerada: {mensagem}")
    # Emite o alerta por voz usando o gTTS
    emitir_alerta_voz(mensagem)

# Executa o sensor inteligente na imagem de teste
processar_sensor_visao('teste.jpg')
```
